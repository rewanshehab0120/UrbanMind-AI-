```python
import os
import urllib
import pandas as pd
import pyodbc
from sqlalchemy import create_engine, types

# 1. إعداد بيانات الاتصال بـ SQL Server
server = r"REWANSHEHAB\SQLEXPRESS"
database = "UrbanMindDB"

# 2. قراءة ملف master_dataset_2.csv
base_dir = r"D:\Courses\MachineLearning\Depi\Final Project\data"
csv_path = os.path.join(base_dir, "master_dataset_2.csv")
df = pd.read_csv(csv_path)

# 3. الاتصال بـ SQL Server
db_conn_str = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(db_conn_str)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# 4. تحديد أنواع الأعمدة النصية لتكون NVARCHAR للغة العربية
dtype_dict = {
    col: types.NVARCHAR(length=255)
    for col in df.select_dtypes(include=["object"]).columns
}

# 5. رفع البيانات مع الحفاظ على اللغة العربية
df.to_sql(
    "master_dataset",
    con=engine,
    if_exists="replace",
    index=False,
    dtype=dtype_dict,
)
print("🚀 Table 'master_dataset' re-created with Arabic support successfully!")
```
