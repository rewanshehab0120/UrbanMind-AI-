```python
# Cell 2 — Imports and project paths

from pathlib import Path

import pandas as pd
import numpy as np
import requests

print("Libraries imported successfully.")
```

```python
# Cell 3 — Create population data folders

BASE_DIR = Path.cwd()

POPULATION_DIR = BASE_DIR / "data" / "population"

RAW_DIR = POPULATION_DIR / "raw"
PROCESSED_DIR = POPULATION_DIR / "processed"
REPORTS_DIR = POPULATION_DIR / "reports"

RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

print("Population folders created successfully.")
print("Population directory:", POPULATION_DIR)
```

```python
# Cell 4 — Test connection to CAPMAS

CAPMAS_URL = "https://censusinfo.capmas.gov.eg/Metadata-Ar-v4.2/index.php/catalog/1723"

try:
    response = requests.get(
        CAPMAS_URL,
        timeout=30
    )

    response.raise_for_status()

    print("CAPMAS connection successful.")
    print("Status code:", response.status_code)
    print("Page size:", len(response.text), "characters")

except Exception as e:
    print("CAPMAS connection failed.")
    print(type(e).__name__)
    print(e)
```

```python
# Cell 5 — Test CAPMAS DNS resolution

import socket

CAPMAS_HOST = "censusinfo.capmas.gov.eg"

try:
    ip_address = socket.gethostbyname(CAPMAS_HOST)

    print("DNS resolution successful.")
    print("CAPMAS IP address:", ip_address)

except Exception as e:
    print("DNS resolution failed.")
    print(type(e).__name__)
    print(e)
```

```python
# Cell 6 — Check population raw folder

from pathlib import Path

files = list(RAW_DIR.iterdir())

if files:
    print("Files found in population/raw:")
    for file in files:
        print("-", file.name)
else:
    print("No population files found yet.")
    print("The raw folder is ready for the population dataset.")
```

```python
# Cell 8 — Download and validate official CAPMAS population PDF

import requests

POPULATION_SOURCE_URL = (
    "https://www.capmas.gov.eg/pdf/new_Pdf/"
    "2021530114249_pop_n.pdf"
)

POPULATION_PDF = RAW_DIR / "cairo_population_2021.pdf"

try:
    response = requests.get(
        POPULATION_SOURCE_URL,
        timeout=60
    )

    response.raise_for_status()

    content = response.content

    # Check that the downloaded file is actually a PDF
    if not content.startswith(b"%PDF"):
        print("Downloaded file is NOT a valid PDF.")
        print("First bytes:", content[:100])

    else:
        with open(POPULATION_PDF, "wb") as f:
            f.write(content)

        file_size_kb = POPULATION_PDF.stat().st_size / 1024

        print("Official CAPMAS PDF downloaded successfully.")
        print("Saved to:", POPULATION_PDF)
        print("File size:", round(file_size_kb, 2), "KB")

except Exception as e:
    print("Download failed.")
    print(type(e).__name__)
    print(e)
```
