```python
import urllib
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sqlalchemy import create_engine

# 1. الاتصال بـ SQL Server وقراءة جدول master_dataset
server = r"REWANSHEHAB\SQLEXPRESS"
database = "UrbanMindDB"

db_conn_str = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(db_conn_str)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# 2. استعلام البيانات مباشرة
df = pd.read_sql("SELECT * FROM master_dataset", con=engine)

# 3. عرض الأبعاد وأول الصفوف
print(f"📊 Dataset Loaded from SQL Server! Shape: {df.shape}")
display(df.head())
```
