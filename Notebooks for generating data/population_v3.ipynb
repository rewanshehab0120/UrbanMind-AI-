```python
from pathlib import Path

# مجلد الـ raw
RAW_DIR = Path(
    r"D:\Courses\MachineLearning\Depi\Final Project\data\population\raw"
)

# البحث عن ملفات PDF داخل المجلد
pdf_files = list(RAW_DIR.glob("*.pdf"))

print("Number of PDF files:", len(pdf_files))

print("\nPDF files found:")
for pdf in pdf_files:
    print(pdf.name)
```

```python
import fitz

# أخذ ملف الـPDF الذي وجدناه في Cell 1
PDF_PATH = pdf_files[0]

# فتح الـPDF
doc = fitz.open(PDF_PATH)

print("PDF path:", PDF_PATH)
print("Number of pages:", len(doc))

print("\nPage inspection:")
for i, page in enumerate(doc):
    text = page.get_text("text")
    images = page.get_images(full=True)

    print(
        f"Page {i + 1}: "
        f"text_chars={len(text)}, "
        f"images={len(images)}"
    )
```

```python
import fitz

# تحويل الصفحة الأولى إلى صورة عالية الدقة
page = doc[0]

zoom = 2
matrix = fitz.Matrix(zoom, zoom)

pix = page.get_pixmap(
    matrix=matrix,
    alpha=False
)

print("Page size:", page.rect)
print("Image size:", pix.width, "x", pix.height)
print("Color space:", pix.colorspace)
print("Alpha:", pix.alpha)
```

```python
# أهمية الخلية:
# تحسين صورة صفحة الـPDF قبل تنفيذ OCR عن طريق تحويلها إلى Grayscale
# ثم زيادة وضوح النص باستخدام Threshold، بهدف تقليل أخطاء قراءة
# النصوص والأرقام العربية الموجودة في جدول بيانات السكان.

import cv2
import numpy as np
import pytesseract
from PIL import Image

# تحويل صورة الصفحة إلى OpenCV
img = np.array(image)

# تحويل الصورة إلى Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# تحويل الصورة إلى أبيض وأسود مع تحسين التباين تلقائيًا
_, thresh = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# تنفيذ OCR باللغة العربية والإنجليزية
text = pytesseract.image_to_string(
    thresh,
    lang="ara+eng",
    config="--psm 6"
)

print(text[:5000])
```
