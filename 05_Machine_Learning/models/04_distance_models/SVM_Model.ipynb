This notebook continues directly from the preprocessing notebook. It uses the prepared `X_train`, `X_test`, `y_train`, and `y_test` objects to train and evaluate a Support Vector Regression model.


## 1. Load the preprocessing workflow


Run the preprocessing notebook first. It sets the target, removes irrelevant features, handles missing values, applies feature engineering and selection, creates a spatial split, and scales input features.

## 2. Confirm model inputs


```python
print(f'Target: {TARGET}')
print(f'Training data: {X_train.shape}')
print(f'Test data: {X_test.shape}')
```

## 3. Define the SVM model


```python
from sklearn.svm import SVR
```

```python
svm_model = SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale')
```

## 4. Train the SVM model


```python
svm_model.fit(X_train, y_train)
```

## 5. Generate predictions


```python
svm_predictions = svm_model.predict(X_test)
```

## 6. Evaluate model performance


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
svm_mae = mean_absolute_error(y_test, svm_predictions)
svm_rmse = mean_squared_error(y_test, svm_predictions) ** 0.5
svm_r2 = r2_score(y_test, svm_predictions)
```

```python
print(f'SVM MAE:  {svm_mae:.4f}')
print(f'SVM RMSE: {svm_rmse:.4f}')
print(f'SVM R²:   {svm_r2:.4f}')
```

## 7. Visual diagnostics


### How closely do SVM predictions match actual values?


```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, svm_predictions, alpha=.55, color='royalblue')
limits = [min(y_test.min(), svm_predictions.min()), max(y_test.max(), svm_predictions.max())]
plt.plot(limits, limits, '--', color='crimson', label='Perfect prediction')
plt.xlabel('Actual value')
plt.ylabel('Predicted value')
plt.title('SVM: actual versus predicted')
plt.legend()
plt.show()
```

### Are SVM residuals centred around zero?


```python
svm_residuals = y_test - svm_predictions
```

```python
sns.histplot(svm_residuals, bins=35, kde=True, color='royalblue')
plt.axvline(0, color='crimson', linestyle='--')
plt.title('SVM residual distribution')
plt.xlabel('Actual - predicted')
plt.show()
```

### Does the SVM error change across target values?


```python
plt.scatter(svm_predictions, svm_residuals, alpha=.55, color='royalblue')
plt.axhline(0, color='crimson', linestyle='--')
plt.xlabel('Predicted value')
plt.ylabel('Residual')
plt.title('SVM residuals versus predictions')
plt.show()
```
