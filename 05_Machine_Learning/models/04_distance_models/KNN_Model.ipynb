# UrbanMind KNN Regression Model

This notebook continues directly from the preprocessing notebook. It uses the prepared `X_train`, `X_test`, `y_train`, and `y_test` objects to train and evaluate a K-Nearest Neighbours regression model.


## 1. Load the preprocessing workflow


Run the preprocessing notebook first. It sets the target, removes irrelevant features, handles missing values, applies feature engineering and selection, creates a spatial split, and scales input features.


```python
%run ../../../urbanmind_eda_preprocessing.ipynb
```

## 2. Confirm model inputs


```python
print(f'Target: {TARGET}')
print(f'Training data: {X_train.shape}')
print(f'Test data: {X_test.shape}')
```

## 3. Define the KNN model


```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
knn_model = KNeighborsRegressor(n_neighbors=15, weights='distance', p=2, n_jobs=-1)
```

## 4. Train the KNN model


```python
knn_model.fit(X_train, y_train)
```

## 5. Generate predictions


```python
knn_predictions = knn_model.predict(X_test)
```

## 6. Evaluate model performance


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
knn_mae = mean_absolute_error(y_test, knn_predictions)
knn_rmse = mean_squared_error(y_test, knn_predictions) ** 0.5
knn_r2 = r2_score(y_test, knn_predictions)
```

```python
print(f'KNN MAE:  {knn_mae:.4f}')
print(f'KNN RMSE: {knn_rmse:.4f}')
print(f'KNN R²:   {knn_r2:.4f}')
```

## 7. Visual diagnostics


### How closely do KNN predictions match actual values?


```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, knn_predictions, alpha=.55, color='darkorange')
limits = [min(y_test.min(), knn_predictions.min()), max(y_test.max(), knn_predictions.max())]
plt.plot(limits, limits, '--', color='crimson', label='Perfect prediction')
plt.xlabel('Actual value')
plt.ylabel('Predicted value')
plt.title('KNN: actual versus predicted')
plt.legend()
plt.show()
```

### Are KNN residuals centred around zero?


```python
knn_residuals = y_test - knn_predictions
```

```python
sns.histplot(knn_residuals, bins=35, kde=True, color='darkorange')
plt.axvline(0, color='crimson', linestyle='--')
plt.title('KNN residual distribution')
plt.xlabel('Actual - predicted')
plt.show()
```

### Does KNN error change across target values?


```python
plt.scatter(knn_predictions, knn_residuals, alpha=.55, color='darkorange')
plt.axhline(0, color='crimson', linestyle='--')
plt.xlabel('Predicted value')
plt.ylabel('Residual')
plt.title('KNN residuals versus predictions')
plt.show()
```
